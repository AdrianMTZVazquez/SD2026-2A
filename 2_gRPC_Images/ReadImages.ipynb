```python
IMAGE_NAME = "n008-2018-08-01-15-16-36-0400__CAM_FRONT__1533151611862404.jpg"

TAM = 2 * 512
array_chunk = []
with open(IMAGE_NAME, "rb") as f:
    cnt = 0
    while True:
        image_data = f.read(TAM)
        array_chunk.append(image_data)
        if not image_data:
            break
        cnt += 1
print(f"Total chunks read: {cnt}")
```

```python
nombre = ""
buf = bytearray()
n = 0
for frag in array_chunk:
    buf.extend(frag)
    n += 1
buf

with open("test--------------------.jpg","wb") as f:
    f.write(buf)
```
